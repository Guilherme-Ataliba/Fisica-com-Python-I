```python
import random
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme()
```

```python
def random_walk(x, time):
    Ts, X = [0], [0]
    for ts in range(1, time + 1):
        Ts.append(ts)
        rand = random.choice(x)
        step = X[ts - 1] + rand
        X.append(step)

    plt.plot(Ts, X)
```

```python
for i in range(100):
    random_walk([-1, 1], 5000)

plt.title("Random walk")
plt.xlabel("Time")
plt.ylabel("Walk")
plt.show()
```
